```python
def extract_evidence(row, baseline):
    evidence = []

    # Traffic
    if row["rate"] > baseline.loc[0.99, "rate"]:
        evidence.append("Traffic rate is unusually high")

    if row["sbytes"] > baseline.loc[0.99, "sbytes"]:
        evidence.append("Source bytes are unusually high")

    if row["dbytes"] > baseline.loc[0.99, "dbytes"]:
        evidence.append("Destination bytes are unusually high")

    # Network load
    if row["sload"] > baseline.loc[0.99, "sload"]:
        evidence.append("Source network load is unusually high")

    if row["dload"] > baseline.loc[0.99, "dload"]:
        evidence.append("Destination network load is unusually high")

    # Loss-related signals
    if row["sloss"] > baseline.loc[0.99, "sloss"]:
        evidence.append("Source loss-related signal is unusually high")

    if row["dloss"] > baseline.loc[0.99, "dloss"]:
        evidence.append("Destination loss-related signal is unusually high")

    # Latency
    if row["tcprtt"] > baseline.loc[0.99, "tcprtt"]:
        evidence.append("TCP round-trip time is unusually high")

    if row["synack"] > baseline.loc[0.99, "synack"]:
        evidence.append("SYN-ACK delay is unusually high")

    if row["ackdat"] > baseline.loc[0.99, "ackdat"]:
        evidence.append("ACK delay is unusually high")

    # Connection activity
    if row["ct_srv_src"] > baseline.loc[0.99, "ct_srv_src"]:
        evidence.append("Source-to-service connection activity is unusually high")

    if row["ct_dst_src_ltm"] > baseline.loc[0.99, "ct_dst_src_ltm"]:
        evidence.append("Destination-source connection activity is unusually high")

    return evidence
```

```python
def correlate_evidence(row, baseline):
    evidence = extract_evidence(row, baseline)

    causes = []

    # Rule 1: High traffic / load
    traffic_signals = [
        "Traffic rate is unusually high",
        "Source bytes are unusually high",
        "Destination bytes are unusually high",
        "Source network load is unusually high",
        "Destination network load is unusually high"
    ]

    traffic_count = sum(
        signal in evidence for signal in traffic_signals
    )

    if traffic_count >= 2:
        causes.append({
            "cause": "High traffic or network load",
            "confidence": "medium",
            "supporting_evidence": [
                signal for signal in evidence
                if signal in traffic_signals
            ]
        })

    # Rule 2: Connection activity
    connection_signals = [
        "Source-to-service connection activity is unusually high",
        "Destination-source connection activity is unusually high"
    ]

    connection_count = sum(
        signal in evidence for signal in connection_signals
    )

    if connection_count >= 1:
        causes.append({
            "cause": "Unusual connection activity",
            "confidence": "medium",
            "supporting_evidence": [
                signal for signal in evidence
                if signal in connection_signals
            ]
        })

    # Rule 3: Latency / transport behavior
    latency_signals = [
        "TCP round-trip time is unusually high",
        "SYN-ACK delay is unusually high",
        "ACK delay is unusually high"
    ]

    latency_count = sum(
        signal in evidence for signal in latency_signals
    )

    if latency_count >= 1:
        causes.append({
            "cause": "Abnormal TCP timing or latency",
            "confidence": "low",
            "supporting_evidence": [
                signal for signal in evidence
                if signal in latency_signals
            ]
        })

    # Rule 4: Loss-related behavior
    loss_signals = [
        "Source loss-related signal is unusually high",
        "Destination loss-related signal is unusually high"
    ]

    loss_count = sum(
        signal in evidence for signal in loss_signals
    )

    if loss_count >= 1:
        causes.append({
            "cause": "Loss-related network behavior",
            "confidence": "low",
            "supporting_evidence": [
                signal for signal in evidence
                if signal in loss_signals
            ]
        })

    return {
        "evidence": evidence,
        "candidate_causes": causes
    }
```
